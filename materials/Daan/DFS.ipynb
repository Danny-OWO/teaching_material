```python
# stack-based DFS solution to find if there's a path from top-left to bottom-right in a grid
import sys

infile = open('in.txt')

r, c = map(int, infile.readline().strip().split())

lst = []
for i in range(r):
    lst.append(list(map(int, infile.readline().strip().split())))

stack = []
stack.append((0, 0))
visited = [[False] * c for _ in range(r)]
while stack:
    current_i, current_j = stack.pop()
    if lst[current_i][current_j] == 1:
        continue
    if current_i == r - 1 and current_j == c - 1:
        print("yes")
        break
    lst[current_i][current_j] = 1
    for di, dj in [(0, 1), (1, 0), (0, -1), (-1, 0)]:
        dr, dc = current_i + di, current_j + dj
        if 0 <= dr < r and 0 <= dc < c and not visited[dr][dc]:
            visited[dr][dc] = True
            stack.append((dr, dc))
print("no")
```
