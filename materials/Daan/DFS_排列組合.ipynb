```python
# Leetcode 17. Letter Combinations of a Phone Number
ans = []
d = {"2": "abc", "3": "def", "4": "ghi", "5": "jkl", "6": "mno", "7": "pqrs", "8": "tuv", "9": "wxyz"}
def f(index, lst, digits):
    if index == len(digits):
        ans.append("".join(lst))
        return
    for i in range(len(d[digits[index]])):
        lst.append(d[digits[index]][i])
        f(index + 1, lst, digits)
        lst.pop()
digits = "23"
f(0, [], digits)
print(ans)
```

```python
# Leetcode 784. Letter Case Permutation

ans = []
s = "a1b2"
def f(index, lst, s):
    if index == len(s):
        ans.append("".join(lst))
        return
    if s[index].isalpha():
        lst.append(s[index].lower())
        f(index + 1, lst, s)
        lst.pop()

        lst.append(s[index].upper())
        f(index + 1, lst, s)
        lst.pop()
    else:
        lst.append(s[index])
        f(index + 1, lst, s)
        lst.pop()
f(0, [], s)
print(ans)
```

### 排列

```python
def f(n, lst, used):
    if len(lst) == n:
        print(lst)
        return

    for i in range(1, n + 1):
        if not used[i]:
            used[i] = True
            lst.append(i)
            f(n, lst, used)
            lst.pop()
            used[i] = False

n = 3
f(n, [], [False] * (n + 1))
```
